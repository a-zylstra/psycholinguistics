```python
import pandas as pd
import numpy as np
```

```python
columns_to_keep = [
    "subjectID"
    , "trial_property"
    , "response"
    , "type"
    , "item"
    , "art_aux"
    , "prep_adv"
    ,"sentence_fragment"
    , "left_label"
    , "pres_order"
    , "...22"
    , "name"
    , "par_no"
]
```

```python
MVRR_df = pd.read_csv("MVRR_data_Oct19_forclass.csv")
# NPS_df = pd.read_csv("NPS_data_Oct19_forclass.csv")
```

```python
MVRR_df = MVRR_df[columns_to_keep]

#TODO when making function, "trial-{}" change it 
MVRR_df = MVRR_df[(MVRR_df["trial_property"] == "trial-MVRR") & (MVRR_df["type"] != "FILLER")]
```

```python
# sanity check, should be true 
(MVRR_df[["subjectID", "item"]].value_counts() == 3).all()
```

```python
MVRR_df["adjusted_response"] = np.where(
    MVRR_df["left_label"].isin(["RR", "S"]),
    100 - MVRR_df["response"],
    MVRR_df["response"]
)
```

```python
starting_count = len(MVRR_df)
print(f"Started with {starting_count} rows...")
MVRR_df = MVRR_df.dropna(subset=["response"]).reset_index(drop=True)
number_dropped = starting_count - len(MVRR_df)
print(f"Ending with {len(MVRR_df)}, dropping {number_dropped} rows.")
```
