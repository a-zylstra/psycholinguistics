```python
import pandas as pd
```

```python
MVRR_df_human = pd.read_csv('../assn1b/MVRR_data_Oct19_forclass.csv')
NPS_df_human = pd.read_csv('../assn1b/NPS_data_Oct19_forclass.csv')
```

```python
MVRR_df_llm = pd.read_csv("llm_MVRR_results.csv")
NPS_df_llm = pd.read_csv("llm_NPS_results.csv")
```

```python
MVRR_df_llm = MVRR_df_human[['item', 'sentence_fragment']].drop_duplicates().merge(MVRR_df_llm, left_on="sentence_fragment", right_on="fragment")
```

```python
NPS_df_llm = NPS_df_human[['item', 'sentence_fragment']].drop_duplicates().merge(NPS_df_llm, left_on="sentence_fragment", right_on="fragment")
```

```python
verb_MVRR = pd.read_csv("../assn1b/MVRR_matrix_verb.csv")
verb_NPS = pd.read_csv("../assn1b/NPS_matrix_verb.csv")
```

```python
verb_MVRR[["verb", "item"]] = verb_MVRR["matrix_verb_itemized"].str.rsplit("_", n=1, expand=True)
```

```python
verb_NPS[["verb", "item"]] = verb_NPS["matrix_verb_itemized"].str.rsplit("_", n=1, expand=True)
```

```python
MVRR_df_llm = verb_MVRR[['item', 'verb']].drop_duplicates().merge(MVRR_df_llm, on="item")
```

```python
NPS_df_llm = verb_NPS[['item', 'verb']].drop_duplicates().merge(NPS_df_llm, on="item")
```
